# Data Normalization

### <pre>1. Starting the PySpark Session

```python
import pyspark
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('tutorial_7').getOrCreate()
```

### <pre>2. Loading the Dataset

```python
import seaborn as sns

df = sns.load_dataset("iris")

df2 = spark.createDataFrame(df)
```

### <pre>3. Vector Assembling

```python
from pyspark.ml.feature import VectorAssembler

featureAssembler = VectorAssembler(inputCols = df2.columns[:4],
                                  outputCol="FeatureVector")

df3 = featureAssembler.transform(df2)

df4 = df3["FeatureVector", "species"]
```

```python
df4.show(10)
```

### <pre>4. Data Normalizing

```python
from pyspark.ml.feature import MinMaxScaler
```

```python
scaler = MinMaxScaler(inputCol = 'FeatureVector',
                     outputCol = 'ScaledFeatures')
```

```python
df = scaler.fit(df4).transform(df4)
```

```python
df.show(10)
```

```python
df6 = df["ScaledFeatures", "species"]
```

```python
df6.show(10)
```
