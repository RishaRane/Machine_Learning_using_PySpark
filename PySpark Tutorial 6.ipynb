# Vector Assembler

```python
import pyspark
```

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("VectorAssesmbler").getOrCreate()
```

```python
spark
```

```python
import seaborn as sns
```

```python
df = sns.load_dataset("iris")
```

```python
df = spark.createDataFrame(df)
```

```python
df.show(10)
```

```python
df.columns[:4]
```

```python
from pyspark.ml.feature import VectorAssembler

featureAssembler = VectorAssembler(inputCols = df.columns[:4], outputCol="FeatureVector")
```

```python
new_df = featureAssembler.transform(df)
```

```python
new_df.show(10)
```

```python
final_df = new_df['FeatureVector', 'species']
```

```python
final_df.show(10)
```

```python
from pyspark.ml.feature import MinMaxScaler

scaler = MinMaxScaler(inputCol='FeatureVector', outputCol='ScaledFeatures')
```

```python
df = scaler.fit(final_df).transform(final_df)
```

```python
df.show(10)
```

```python
df = df['ScaledFeatures', 'species']
```

```python
df.show(10)
```
